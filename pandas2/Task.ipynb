```python
#1st clean the data [remove the duplicates]
df2 = df.copy()

df2 = df2.drop_duplicates()
df2 = df2.fillna(0)
```

```python
## Grouping & Aggregation of Data
df
```

```python
df.groupby("country")["income"].mean()
df.groupby("country")["income"].min()
df.groupby("country")["income"].max()

#this all is used to help apply in the specific column
df.groupby("gender")["income"].mean()
df.groupby("gender")["income"].max()
```

```python
# to apply in all the column we use agg()
df.groupby("country")["income"].agg(["mean", "min" , "max"])
```

```python
df.groupby("country")["income"].agg(avg_salary="mean", min_salary ="min", max_salary ="max")
```

```python
#we can use different aggregate function in column
df.groupby("country").agg({
    "income" : "mean",
    "age" : "mean"
})
df.groupby("country").agg({
    "income" : "max",
    "age" : "mean"
})
```
