```python
#implementaion of neural network
import torch
import torch.nn as nn
```

```python
#to get exact output we can use {torch.manual_seed(42)}
inputs = torch.tensor([1.0, 2.0, 3.0])
```

```python
neuron = nn.Linear(in_features=3, out_features=1)
```

```python
output=neuron(inputs)
```
