```python
def clean_data(data):
    text_to_num = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5}
    cleaned_data = []
    unique_users = set()

    for user in data:
        # Clean rating
        raw_rating = str(user["rating"]).strip().lower()
        if raw_rating in text_to_num:
            raw_rating = text_to_num[raw_rating]
        user["rating"] = raw_rating

        # Handle missing age
        raw_age = user.get("age")
        if raw_age is None:
            user["age"] = None

        # Deduplication (case + whitespace normalized)
        clean_name = user["name"].strip().lower()
        if clean_name in unique_users:
            continue
        
        unique_users.add(clean_name)
        cleaned_data.append(user)

    return cleaned_data
```

```python
data=clean_data(data)
```

```python
#get meaningful insights
def get_insights(data):

    #avg rating
    tot_rating=0
    for user in data:
        tot_rating += float(user["rating"])

    print(f"avg rating ={tot_rating/len(data)}")

    #percentage of users with poor rating
    poor_rating =0
    for user in data:
        if(float(user["rating"]) <3):
            poor_rating +=1
    print(f"% of user with poor rating = {poor_rating/len(data)*100}%")
    
    
```

```python
get_insights(data)
```

```python
#build recomendation feature
def get_recommendations(data):
    recommendations = []

    for user in data:
        curr_recomm = {}
        curr_recomm["name"] = user["name"]

        if float(user["rating"]) >= 4:
            curr_recomm["brand"] = "Apple"
        else:
            curr_recomm["brand"] = "Samsung"

        recommendations.append(curr_recomm)

    return recommendations

print(get_recommendations(data))
```

```python
def clean_missing(data):
    for user in data:
        # handle missing age
        if "age" not in user or user["age"] is None:
            user["age"] = None

        # handle missing feedback
        if "feedback" not in user or user["feedback"] is None:
            user["feedback"] = None

        # handle missing rating
        if "rating" not in user or user["rating"] is None:
            user["rating"] = None
```
