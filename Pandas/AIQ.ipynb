```python
lowest = df.loc[df["aqi"].idxmin(), ["city","aqi"]]
```

```python
print(lowest)
```

```python
highest_per_city = df.loc[df.groupby("city")["aqi"].idxmax()]
print(highest_per_city[["city", "aqi","temperature"]])
```

```python
result = highest_per_city[["city","aqi","temperature"]]
```

```python
result.to_excel("highest_aqi.xlsx", index=False)
```
