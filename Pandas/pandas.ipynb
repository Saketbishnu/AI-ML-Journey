```python
#data analysis and manipulation
import pandas as pd
```

```python
info ={
    "Name" : ["saket", "bhavya", "rahul"],
    "CGPA" : ["8.1","7,5" ,"8.5"]

}
df = pd.DataFrame(info)
print(df)

print(df.iloc[0])
df.iloc[0,0]
df.describe() 
```

```python
s = pd.Series([1,2,3,4])
print(s)
#index
print(s[0])
print(s[2])
```

```python
#we can assign label values
s = pd.Series([23,24,25,26], index=["Adam", "Eve", "Bob", "charlie"])
print(s)
print(s.index)
```

```python
#vectorization
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([10,20,30,40,50])

print(s1+s2)
s1[0] =100
print(s1)
```

```python
#dataframe
info ={
    "Name" :["Adam", "Eve", "Bob"],
    "Age" : [23,24,25],
    "GPA" : [9.5, 8.6 , 7.2]
}
df= pd.DataFrame(info)
print(df,type(df))

print(df.index)
print(df.columns)
```
